## Dimesnional Arrays in NumPy

- `ndarray` - n-dimensional-array
    - `0-D array` - single value
    - `1-D array` - list (one direction)
    - `2-D array` - table (rows & columns)
    - `3-D array` - multiple 2D tables stacked togetehr

```python
import numpy as np
```

```python
# 0-D arrray

a0 = np.array(5)
print('value:', a0)
print('dimesnsion:', a0.ndim)
```

```python
# 1-D array

a1 = np.array([1, 2, 3])
print('value:', a1)
print('dimension:', a1.ndim)
```

```python
# 2-D array

a2 = np.array([[1, 2, 3], 
              [4, 5, 6]])

print('value:', a2)
print('dimension:', a2.ndim)
```

```python
# 3-D array

a3 = np.array([[[1, 2], [3, 4]], 
               [[5, 6], [7, 8]]])

print('value:', a3)
print('dimension:', a3.ndim)
```

```python
#vectorised operations

arr = np.array([1, 2, 3, 4])
print('additon:', arr + 10)
print('multiplication:', arr*2)
print('to the power of:', arr**2)
```

```python
arr = np.array([1, 4, 9, 16])
print('Square root:', np.sqrt(arr))
print('Sum:', np.sum(arr))
print('Mean:', np.mean(arr))
print('Standard Devitation:', np.std(arr))
```

```python
a = np.array([1, 2])
b = np.array([3, 4])
c = np.concatenate([a, b])
d = np.vstack((a, b))

print(a)
print(b)
print('Joined the two arrays:', c)
print('Created a table:', d)
```

```python
arr = np.arange(1, 10)
print('First arr:', arr)

arr1 = arr.reshape(3, 3)
print('Reshaped array:', arr1)

arr2 = arr*5
print('Multiply original arr by 5:', arr2)

row_means = np.mean(arr1, axis=1)
print('Mean of rows:', row_means)
```
