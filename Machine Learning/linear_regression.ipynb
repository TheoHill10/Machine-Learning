# Linear Regression

### Linear Regression Formula

`y = m*x + b`
- `y` - output or the value we are predicting
- `x` - input or the feature used to make predictions
- `m` - slope, tells us how muc `y` changes when `x` increase
- `b` - intercept, where the line croses the y-axis when `x` = 0

`m = 2` - for evey additional hour studied the student is expected to score 2 points higher <br>
`b = 1.5` - for students who doesn't study at all they are expected to score 1.5

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

`X` - features (always 2D array)
- rows = samples
- columns = features

```python
#hours of studying
X = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])

#score on tests
y = np.array([2, 3, 5, 4, 6, 5, 7, 8])
```

```python
df = pd.DataFrame({'Hours_Studied': X.flatten(), 'Grade': y})

plt.figure(figsize=(8,4))
plt.scatter(df['Hours_Studied'], df['Grade'], color = 'seagreen')
plt.title('Study Hours vs Grade')
plt.xlabel('Hours Studies')
plt.ylabel('Grade')
plt.grid(True)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)
```

```python
model.predict([[9]])
```

```python
plt.figure(figsize=(8, 4))
plt.scatter(X, y, color='purple', label='Actual Data')

x_line = np.linspace(0, 9, 100).reshape(-1, 1)
y_line = model.predict(x_line)

plt.plot(x_line, y_line, color= 'red', label='Model Data')
plt.title('Linear Regression: Study Hours vs Grade')
plt.xlabel('Hours studied')
plt.ylabel('Grade')
plt.legend()
plt.grid()
```

```python
print('Slope (m):', model.coef_)
print('Intercept (b):', model.intercept_)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
print(X)
print(y)
```

```python
y_pred = model.predict(X)
y_pred
```

```python
print('MSE:', mean_squared_error(y, y_pred))
print('R2 Score:', r2_score(y, y_pred))
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='purple', label='Actual Data')
plt.scatter(X, y_pred, color='red', label='Predictd Data')
plt.title('Actual vs Predicted Grades')
plt.grid()
```
